## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
My lane is Content Refresh Prioritisation. The task is to score each page on how urgently it needs a refresh, then rank pages from highest to lowest priority. This is a scoring problem — not binary classification — because the output is a ranked list, not just "refresh" or "don't refresh." A fixed rule like "refresh pages older than 6 months" ignores actual performance signals like CTR drop and position decline, which is why ML beats it here.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
The target column is trend_direction — whether a page is trending "down", "stable", or "up" in impressions. This is an observed outcome from the data, not a manually defined rule. I will use "down" as the positive class — pages that are declining are the ones most worth refreshing. This is a proxy because we cannot directly observe "needs refresh" — we observe performance decline as a signal for it.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
A fixed rule like "refresh pages older than 90 days" ignores actual performance. A page can be old but still ranking well. A page can be new but already declining. ML combines multiple signals — position tier, CTR drop, content age, impression trend — and learns which combination actually predicts decline. No single threshold captures this interaction. That is what makes it an ML problem, not a rule problem.
```
