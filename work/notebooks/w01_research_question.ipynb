## 2. The question: decision, action, cost of a wrong call

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
Question: Which pages are most worth refreshing based on their current search performance signals?

Decision: A content editor decides which pages to update this week — they cannot refresh all 50,000+ pages, so they need a ranked list.

Action: The editor opens the top-ranked pages from the model's output and schedules them for a content update.

Cost of a wrong call: If the model ranks the wrong pages as high priority, the editor wastes time refreshing pages that did not need it — and the pages that actually needed updating keep declining. The cost is wasted editor time and continued traffic loss.
```

## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(f"Total pages: {len(df)}")
print(f"\nPages with declining trend: {(df['trend_direction'] == 'down').sum()}")
print(f"Declining pages %: {(df['trend_direction'] == 'down').mean()*100:.1f}%")
print(f"\nMean CTR by position tier:")
print(df.groupby('position_tier')['ctr'].mean().sort_values(ascending=False).round(4))
```
